### 5) Análisis de complejidad **[15 pts]**

Justificación

Max-Heap (MAX-HEAPIFY, BUILD-MAX-HEAP, EXTRACT-MAX e inserción):

MAX-HEAPIFY compara un nodo con sus dos hijos y desciende a lo sumo hasta una hoja, tomando $O(\log n)$.

BUILD-MAX-HEAP recorre los nodos internos desde $\lfloor n/2 \rfloor - 1$ hasta $0$ ejecutando MAX-HEAPIFY; mediante el análisis ajustado por alturas de niveles, su complejidad es $\Theta(n)$.

EXTRACT-MAX reemplaza la raíz por el último elemento en $O(1)$ y restaura el heap con MAX-HEAPIFY en $O(\log n)$.La inserción de T07 agrega el elemento al final en $O(1)$ y realiza sift-up hasta la raíz en a lo sumo $O(\log n)$.

- Función T[n]

T[n] = $an^4 $+ $bn^3 $+$cn^2 $+ dn + $e$ $log2$ n + $f$

Complejidad total: $O(n^4)$ esto debido al ordenamiento $O(n^2)$ aristas con selection sort.

```python
%pip install matplotlib
import math
import matplotlib.pyplot as plt

# ==========================================
# PUNTO 1: COLA DE PRIORIDAD (MAX-HEAP)
# ==========================================
# Cada tarea es una lista: [codigo, nombre, prioridad]
tareas = [
    ["T01", "Configurar servidor", 6],
    ["T02", "Reparar sensor", 10],
    ["T03", "Actualizar software", 4],
    ["T04", "Revisar cámara", 8],
    ["T05", "Generar respaldo", 3],
    ["T06", "Configurar alarma", 7]
]

def imprimir_arreglo(lista):
    for t in lista:
        print(t[0], "-", t[1], "| Prioridad:", t[2])

def imprimir_arbol(lista):
    n = len(lista)
    nivel = 0
    i = 0
    while i < n:
        cantidad = 2 ** nivel
        linea = ""
        for _ in range(cantidad):
            if i < n:
                linea = linea + lista[i][0] + "(" + str(lista[i][2]) + ")   "
                i = i + 1
        print("Nivel", nivel, ":", linea)
        nivel = nivel + 1

def max_heapify(lista, n, i):
    mayor = i
    izq = 2 * i + 1
    der = 2 * i + 2

    if izq < n and lista[izq][2] > lista[mayor][2]:
        mayor = izq

    if der < n and lista[der][2] > lista[mayor][2]:
        mayor = der

    if mayor != i:
        temp = lista[i]
        lista[i] = lista[mayor]
        lista[mayor] = temp
        max_heapify(lista, n, mayor)

def build_max_heap(lista):
    n = len(lista)
    for i in range(n // 2 - 1, -1, -1):
        max_heapify(lista, n, i)

print("--- ARREGLO INICIAL ---")
imprimir_arreglo(tareas)

build_max_heap(tareas)

print("\n--- MAX-HEAP RESULTANTE ---")
imprimir_arreglo(tareas)

print("\n--- ARBOL DEL HEAP ---")
imprimir_arbol(tareas)

# ==========================================
# PUNTO 2: PROCESAR E INSERTAR TAREAS
# ==========================================
def extract_max(lista):
    n = len(lista)
    if n == 0:
        return None
    if n == 1:
        return lista.pop()

    maximo = lista[0]
    lista[0] = lista.pop()
    max_heapify(lista, len(lista), 0)
    return maximo

def insertar_tarea(lista, nueva):
    lista.append(nueva)
    i = len(lista) - 1
    while i > 0:
        padre = (i - 1) // 2
        if lista[i][2] > lista[padre][2]:
            temp = lista[i]
            lista[i] = lista[padre]
            lista[padre] = temp
            i = padre
        else:
            break

for num in range(1, 3):
    sacada = extract_max(tareas)
    print("\n--- EXTRACCION", num, "---")
    print("Tarea procesada:", sacada[0], "-", sacada[1], "(Prioridad:", sacada[2], ")")
    print("Heap resultante:")
    imprimir_arreglo(tareas)

nueva_tarea = ["T07", "Falla en sistema contra incendios", 12]
insertar_tarea(tareas, nueva_tarea)

print("\n--- HEAP DESPUES DE INSERTAR T07 ---")
imprimir_arreglo(tareas)
print("Arbol actualizado:")
imprimir_arbol(tareas)

# ==========================================
# PUNTO 3: DISPOSITIVOS Y DISTANCIAS
# ==========================================
# Cada dispositivo es: [nombre, id, x, y]
dispositivos = [
    ["Servidor", "A", 2, 2],
    ["Cámara",   "B", 7, 3],
    ["Sensor",   "C", 4, 7],
    ["Alarma",   "D", 10, 8],
    ["Router",   "E", 13, 3],
    ["Panel",    "F", 15, 7]
]

plt.figure(figsize=(8, 6))
for d in dispositivos:
    plt.scatter(d[2], d[3], s=100)
    plt.text(d[2] + 0.3, d[3] + 0.3, d[1] + " (" + d[0] + ")")

plt.title("Posicion de los dispositivos")
plt.xlabel("X")
plt.ylabel("Y")
plt.xlim(0, 17)
plt.ylim(0, 10)
plt.grid(True)
plt.show()

# Cada arista es: [id1, id2, distancia]
aristas = []
n_disp = len(dispositivos)

for i in range(n_disp):
    for j in range(i + 1, n_disp):
        d1 = dispositivos[i]
        d2 = dispositivos[j]
        dx = d2[2] - d1[2]
        dy = d2[3] - d1[3]
        distancia = math.sqrt(dx * dx + dy * dy)
        aristas.append([d1[1], d2[1], distancia])

print("\n--- DISTANCIAS ENTRE DISPOSITIVOS ---")
for a in aristas:
    print(f"{a[0]} - {a[1]} : {a[2]:.2f}")

# ==========================================
# PUNTO 4: KRUSKAL MANUAL (MST)
# ==========================================
# Ordenar aristas de menor a mayor peso (selection sort básico)
aristas_ordenadas = []
for a in aristas:
    aristas_ordenadas.append(a)

m = len(aristas_ordenadas)
for i in range(m):
    menor = i
    for j in range(i + 1, m):
        if aristas_ordenadas[j][2] < aristas_ordenadas[menor][2]:
            menor = j
    temp = aristas_ordenadas[i]
    aristas_ordenadas[i] = aristas_ordenadas[menor]
    aristas_ordenadas[menor] = temp

# Estructura simple para evitar ciclos (cada nodo guarda su grupo)
grupo = {}
for d in dispositivos:
    nodo_id = d[1]
    grupo[nodo_id] = nodo_id

def buscar_grupo(grupo, nodo):
    while grupo[nodo] != nodo:
        nodo = grupo[nodo]
    return nodo

def unir_grupos(grupo, u, v):
    gu = buscar_grupo(grupo, u)
    gv = buscar_grupo(grupo, v)
    if gu == gv:
        return False
    grupo[gv] = gu
    return True

mst = []
total_cable = 0.0

for a in aristas_ordenadas:
    u = a[0]
    v = a[1]
    peso = a[2]
    if unir_grupos(grupo, u, v):
        mst.append([u, v, peso])
        total_cable = total_cable + peso
        if len(mst) == len(dispositivos) - 1:
            break

print("\n--- CONEXIONES DEL MST ---")
for c in mst:
    print(f"{c[0]} - {c[1]} : {c[2]:.2f}")

print(f"Longitud total de cable: {total_cable:.2f}")

# ==========================================
# PUNTO 5: GRAFICA FINAL DEL MST
# ==========================================
coords = {}
for d in dispositivos:
    coords[d[1]] = (d[2], d[3])

plt.figure(figsize=(8, 6))

for c in mst:
    x1, y1 = coords[c[0]]
    x2, y2 = coords[c[1]]
    plt.plot([x1, x2], [y1, y2], linewidth=2)
    mx = (x1 + x2) / 2
    my = (y1 + y2) / 2
    plt.text(mx, my + 0.25, f"{c[2]:.2f}")

for d in dispositivos:
    plt.scatter(d[2], d[3], s=100)
    plt.text(d[2] + 0.3, d[3] + 0.3, d[1] + " (" + d[0] + ")")

plt.title(f"MST - Longitud total de cable: {total_cable:.2f}")
plt.xlabel("X")
plt.ylabel("Y")
plt.xlim(0, 17)
plt.ylim(0, 10)
plt.grid(True)
plt.show()
```

```python
import math

print("\n" + "=" * 60)
print(" TEST APLICADOS  ")
print("=" * 60)

# PRUEBA 1: BUILD-MAX-HEAP
tareas_prueba = [
    ["T01", "Configurar servidor", 6],
    ["T02", "Reparar sensor", 10],
    ["T03", "Actualizar software", 4],
    ["T04", "Revisar cámara", 8],
    ["T05", "Generar respaldo", 3],
    ["T06", "Configurar alarma", 7]
]

build_max_heap(tareas_prueba)
print("\nPrueba 1: BUILD-MAX-HEAP con tareas T01-T06")
print("Raíz obtenida:", tareas_prueba[0][0], "Prioridad:", tareas_prueba[0][2])

assert len(tareas_prueba) == 6
assert tareas_prueba[0][0] == "T02" and tareas_prueba[0][2] == 10

for idx in range(len(tareas_prueba)):
    izq = 2 * idx + 1
    der = 2 * idx + 2
    if izq < len(tareas_prueba):
        assert tareas_prueba[idx][2] >= tareas_prueba[izq][2]
    if der < len(tareas_prueba):
        assert tareas_prueba[idx][2] >= tareas_prueba[der][2]

print("✔ Prueba 1 Superada.")

# PRUEBA 2: PRIMERA EXTRACT-MAX
ext_1 = extract_max(tareas_prueba)
assert ext_1[0] == "T02" and ext_1[2] == 10
assert len(tareas_prueba) == 5
assert tareas_prueba[0][2] == 8
print("✔ Prueba 2 Superada.")

# PRUEBA 3: SEGUNDA EXTRACT-MAX
ext_2 = extract_max(tareas_prueba)
assert ext_2[0] == "T04" and ext_2[2] == 8
assert len(tareas_prueba) == 4
assert tareas_prueba[0][2] == 7
print("✔ Prueba 3 Superada.")

# PRUEBA 4: INSERTAR T07 (PRIORIDAD 12)
insertar_tarea(tareas_prueba, ["T07", "Falla en sistema contra incendios", 12])
assert len(tareas_prueba) == 5
assert tareas_prueba[0][0] == "T07" and tareas_prueba[0][2] == 12
print("✔ Prueba 4 Superada.")

# PRUEBA 5: DISTANCIAS ENTRE DISPOSITIVOS
disp_prueba = [
    ["Servidor", "A", 2, 2],
    ["Cámara",   "B", 7, 3],
    ["Sensor",   "C", 4, 7],
    ["Alarma",   "D", 10, 8],
    ["Router",   "E", 13, 3],
    ["Panel",    "F", 15, 7]
]

aristas_prueba = []
for i in range(len(disp_prueba)):
    for j in range(i + 1, len(disp_prueba)):
        d1 = disp_prueba[i]
        d2 = disp_prueba[j]
        dist = math.sqrt((d2[2] - d1[2]) ** 2 + (d2[3] - d1[3]) ** 2)
        aristas_prueba.append([d1[1], d2[1], dist])

assert len(aristas_prueba) == 15
dist_ab_esperada = math.sqrt((7 - 2) ** 2 + (3 - 2) ** 2)
dist_ab_obtenida = None
for a in aristas_prueba:
    if a[0] == "A" and a[1] == "B":
        dist_ab_obtenida = a[2]
        break

assert dist_ab_obtenida is not None and abs(dist_ab_obtenida - dist_ab_esperada) < 1e-9
print("✔ Prueba 5 Superada.")

# PRUEBA 6: ORDENAMIENTO DE ARISTAS
aristas_ord = []
for a in aristas_prueba:
    aristas_ord.append(a)

for i in range(len(aristas_ord)):
    menor = i
    for j in range(i + 1, len(aristas_ord)):
        if aristas_ord[j][2] < aristas_ord[menor][2]:
            menor = j
    aristas_ord[i], aristas_ord[menor] = aristas_ord[menor], aristas_ord[i]

for idx in range(len(aristas_ord) - 1):
    assert aristas_ord[idx][2] <= aristas_ord[idx + 1][2]
print("✔ Prueba 6 Superada.")

# PRUEBA 7: KRUSKAL MST
grupo_p = {}
for d in disp_prueba:
    grupo_p[d[1]] = d[1]

mst_p = []
total_p = 0.0
for a in aristas_ord:
    if unir_grupos(grupo_p, a[0], a[1]):
        mst_p.append(a)
        total_p = total_p + a[2]
        if len(mst_p) == len(disp_prueba) - 1:
            break

assert len(mst_p) == 5
assert total_p > 0

conectados = set()
for c in mst_p:
    conectados.add(c[0])
    conectados.add(c[1])

assert conectados == {"A", "B", "C", "D", "E", "F"}
print("✔ Prueba 7 Superada.")

print("\n" + "=" * 60)
print("   ¡TODAS LAS PRUEBAS DE ENUNCIADO PASARON EXITOSAMENTE!")
print("=" * 60)

import math

# ==========================================
# PRUEBAS PROPIAS
# ==========================================

print("\n" + "=" * 60)
print("     INICIANDO PRUEBAS PROPIAS Y VALIDACIÓN")
print("=" * 60)


# ==========================================
# 1. CONJUNTO PROPIO DE TAREAS Y DISPOSITIVOS
# ==========================================

# Cada tarea: [codigo, nombre, prioridad]
tareas_propias = [
    ["P01", "Reiniciar switch", 5],
    ["P02", "Corte de energía general", 15],
    ["P03", "Limpiar logs", 2],
    ["P04", "Actualizar firmware", 9],
    ["P05", "Verificar UPS", 11]
]

# Cada dispositivo: [nombre, id, x, y]
dispositivos_propios = [
    ["Nodo1", "N1", 0, 0],
    ["Nodo2", "N2", 3, 4],  # Distancia 5.0 exacta a N1
    ["Nodo3", "N3", 3, 0],  # Distancia 3.0 a N1 y 4.0 a N2
    ["Nodo4", "N4", 8, 0]   # Distancia 5.0 a N3
]


# ==========================================
# PRUEBA PROPIA 1:
# HEAP CON PRIORIDAD DUPLICADA Y RAÍZ MÁXIMA
# ==========================================

tareas_duplicadas = [
    ["D01", "Tarea A", 9],
    ["D02", "Tarea B", 9],
    ["D03", "Tarea C", 14],
    ["D04", "Tarea D", 4]
]

build_max_heap(tareas_duplicadas)

print("\nPrueba Propia 1: Max-Heap con prioridades repetidas")
print("Raíz obtenida:", tareas_duplicadas[0][0], "Prioridad:", tareas_duplicadas[0][2])

assert tareas_duplicadas[0][0] == "D03" and tareas_duplicadas[0][2] == 14, (
    f"Fallo: Se esperaba raíz D03 (14), se obtuvo {tareas_duplicadas[0]}."
)

for idx in range(len(tareas_duplicadas)):
    izq = 2 * idx + 1
    der = 2 * idx + 2
    if izq < len(tareas_duplicadas):
        assert tareas_duplicadas[idx][2] >= tareas_duplicadas[izq][2], (
            f"Fallo: Padre {idx} tiene menor prioridad que hijo izquierdo {izq}."
        )
    if der < len(tareas_duplicadas):
        assert tareas_duplicadas[idx][2] >= tareas_duplicadas[der][2], (
            f"Fallo: Padre {idx} tiene menor prioridad que hijo derecho {der}."
        )

print("✔ Prueba Propia 1 Superada: El heap maneja prioridades repetidas sin romper el orden.")


# ==========================================
# PRUEBA PROPIA 2:
# VACIAR EL HEAP POR COMPLETO CON EXTRACT-MAX
# ==========================================

copia_tareas_propias = []
for t in tareas_propias:
    copia_tareas_propias.append([t[0], t[1], t[2]])

build_max_heap(copia_tareas_propias)

extraidas_ordenadas = []
while len(copia_tareas_propias) > 0:
    sacada = extract_max(copia_tareas_propias)
    extraidas_ordenadas.append(sacada[2])

print("\nPrueba Propia 2: Extracción secuencial completa")
print("Prioridades extraídas:", extraidas_ordenadas)

assert extraidas_ordenadas == [15, 11, 9, 5, 2], (
    f"Fallo: Las extracciones no salieron en orden descendente [15, 11, 9, 5, 2]. "
    f"Obtenido: {extraidas_ordenadas}"
)

assert extract_max(copia_tareas_propias) is None, (
    "Fallo: Extraer de un heap vacío debería retornar None."
)

print("✔ Prueba Propia 2 Superada: Extracción completa en orden descendente y manejo de heap vacío.")


# ==========================================
# PRUEBA PROPIA 3:
# INSERCIÓN DE TAREA CON BAJA PRIORIDAD Y OTRA MUY ALTA
# ==========================================

heap_insercion = [
    ["X01", "Chequeo", 8],
    ["X02", "Ajuste", 5]
]
build_max_heap(heap_insercion)

insertar_tarea(heap_insercion, ["X03", "Mantenimiento menor", 1])
assert heap_insercion[0][2] == 8, (
    "Fallo: Insertar prioridad 1 no debería cambiar la raíz de prioridad 8."
)

insertar_tarea(heap_insercion, ["X04", "Alerta crítica", 20])
assert heap_insercion[0][0] == "X04" and heap_insercion[0][2] == 20, (
    "Fallo: Insertar prioridad 20 debería colocar a X04 en la raíz."
)

print("\nPrueba Propia 3: Inserciones de baja y alta prioridad")
print("Nueva raíz tras insertar X04:", heap_insercion[0][0], heap_insercion[0][2])
print("✔ Prueba Propia 3 Superada: Inserciones en ambos extremos funcionan correctamente.")


# ==========================================
# PRUEBA PROPIA 4:
# DISTANCIA CONOCIDA (TRIÁNGULO 3-4-5)
# ==========================================

aristas_propias = []
for i in range(len(dispositivos_propios)):
    for j in range(i + 1, len(dispositivos_propios)):
        d1 = dispositivos_propios[i]
        d2 = dispositivos_propios[j]
        dx = d2[2] - d1[2]
        dy = d2[3] - d1[3]
        dist = math.sqrt(dx * dx + dy * dy)
        aristas_propias.append([d1[1], d2[1], dist])

print("\nPrueba Propia 4: Distancia conocida entre N1(0,0) y N2(3,4)")

dist_n1_n2 = None
for a in aristas_propias:
    if a[0] == "N1" and a[1] == "N2":
        dist_n1_n2 = a[2]
        break

assert dist_n1_n2 is not None and abs(dist_n1_n2 - 5.0) < 1e-9, (
    f"Fallo: La distancia entre (0,0) y (3,4) debe ser 5.0, se obtuvo {dist_n1_n2}."
)

print("✔ Prueba Propia 4 Superada: Distancia euclidiana exacta verificada (5.00).")


# ==========================================
# PRUEBA PROPIA 5:
# MST EN GRAFO PROPIO CON PESO CONOCIDO
# ==========================================

aristas_ord_propias = []
for a in aristas_propias:
    aristas_ord_propias.append([a[0], a[1], a[2]])

m_p = len(aristas_ord_propias)
for i in range(m_p):
    menor = i
    for j in range(i + 1, m_p):
        if aristas_ord_propias[j][2] < aristas_ord_propias[menor][2]:
            menor = j
    temp = aristas_ord_propias[i]
    aristas_ord_propias[i] = aristas_ord_propias[menor]
    aristas_ord_propias[menor] = temp

grupo_propio = {}
for d in dispositivos_propios:
    grupo_propio[d[1]] = d[1]

mst_propio = []
total_propio = 0.0

for a in aristas_ord_propias:
    if unir_grupos(grupo_propio, a[0], a[1]):
        mst_propio.append(a)
        total_propio = total_propio + a[2]
        if len(mst_propio) == len(dispositivos_propios) - 1:
            break

# Para N1(0,0), N2(3,4), N3(3,0), N4(8,0):
# Aristas mínimas sin ciclos: N1-N3 (3.0), N3-N2 (4.0), N3-N4 (5.0) -> Total = 12.0
print("\nPrueba Propia 5: MST sobre 4 nodos propios")
print("Cantidad de conexiones:", len(mst_propio), "| Longitud total:", total_propio)

assert len(mst_propio) == len(dispositivos_propios) - 1, (
    f"Fallo: Para 4 dispositivos se esperaban 3 conexiones, hay {len(mst_propio)}."
)

assert abs(total_propio - 12.0) < 1e-9, (
    f"Fallo: Se esperaba longitud total 12.0, se obtuvo {total_propio}."
)

print("✔ Prueba Propia 5 Superada: Kruskal eligió las 3 aristas mínimas sin ciclos (total 12.00).")


# ==========================================
# RESULTADO FINAL
# ==========================================

print("\n" + "=" * 60)
print("   ¡TODAS LAS PRUEBAS PROPIAS PASARON EXITOSAMENTE!")
print("=" * 60)
```
