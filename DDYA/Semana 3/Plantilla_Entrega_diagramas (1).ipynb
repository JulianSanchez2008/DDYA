## 7) Análisis de complejidad **[15 pts]**

1. Registro de productos

Para registrar n productos, el programa utiliza un ciclo que se ejecuta una vez por cada producto. Por lo tanto, el número de operaciones aumenta proporcionalmente con la cantidad de productos.

La función de tiempo puede expresarse como:

T(n) = cn + d

Por lo tanto:

T(n) = O(n)

Tanto en el mejor como en el peor caso se deben registrar los n productos, por lo que:

Mejor caso: O(n)
Peor caso: O(n)


2. Merge Sort

El algoritmo Merge Sort divide el arreglo en dos subproblemas de aproximadamente la mitad del tamaño.

Su función de recurrencia es:

T(n) = 2T(n/2) + O(n)

Donde:

a = 2: se generan 2 subproblemas.
b = 2: cada subproblema tiene la mitad del tamaño.
f(n) = O(n): la combinación de las dos mitades requiere recorrer los elementos.

Aplicando el Teorema Maestro:

n^(log_b a) = n^(log₂ 2) = n

Como:

f(n) = Θ(n)

se cumple el Caso 2 del Teorema Maestro.

Por lo tanto:

T(n) = Θ(n log n)

En Merge Sort, tanto el mejor como el peor caso mantienen la misma complejidad, ya que el algoritmo siempre divide el arreglo y posteriormente realiza la combinación:

Mejor caso: Θ(n log n)
Peor caso: Θ(n log n)


3. Búsqueda Binaria

La búsqueda binaria funciona sobre un arreglo ordenado. En cada iteración se compara el elemento buscado con el elemento central y se descarta aproximadamente la mitad de los elementos restantes.

La cantidad de elementos se reduce de la siguiente manera:

n → n/2 → n/4 → n/8 → ...

Después de k iteraciones:

n / 2^k = 1

Despejando k:

k = log₂(n)

Por lo tanto, la función de tiempo puede expresarse como:

T(n) = T(n/2) + O(1)

y su complejidad es:

O(log n)

Mejor caso

El mejor caso ocurre cuando el elemento buscado se encuentra exactamente en la posición central en la primera comparación.

T(n) = Θ(1)

Mejor caso: Θ(1)

Peor caso

El peor caso ocurre cuando el elemento no se encuentra o se encuentra después de realizar el máximo número de divisiones del arreglo.

En cada paso se reduce el problema a la mitad:

n → n/2 → n/4 → n/8 → ... → 1

Peor caso: Θ(log n)

```python
#Merge sort Julian Sanchez y Anderson Varela
def merge_sort(lista):
    if len(lista) <= 1:
        return lista

    mitad = len(lista) // 2

    iz = merge_sort(lista[:mitad])
    de = merge_sort(lista[mitad:])

    return merge(iz, de)


def merge(izquierda, derecha):
    resultado = []

    i = 0
    j = 0

    while i < len(izquierda) and j < len(derecha):

        if izquierda[i] < derecha[j]:
            resultado.append(izquierda[i])
            i += 1

        else:
            resultado.append(derecha[j])
            j += 1

    while i < len(izquierda):
        resultado.append(izquierda[i])
        i += 1

    while j < len(derecha):
        resultado.append(derecha[j])
        j += 1

    return resultado

def productos(n,lista):
      vc = 0
      while vc < n:
            producto = int(input(f"Ingrese codigo del producto{vc + 1}"))
            lista.append(producto)
            vc+=1
      return lista
def busqueda_binaria(lista, objetivo):
    izquierda = 0
    derecha = len(lista) - 1

    while izquierda <= derecha:
        medio = (izquierda + derecha) // 2

        if lista[medio] == objetivo:
            return True

        elif lista[medio] < objetivo:
            izquierda = medio + 1

        else:
            derecha = medio - 1

    return False
def main():
      n = int(input("Ingrese la cantidad de productos: "))
      lista = []

      productos(n, lista)

      lista_ordenada = merge_sort(lista)

      print("Ordenado:", lista_ordenada)

      codigo = int(input("Ingrese el código que desea buscar: "))

      encontrado = busqueda_binaria(lista_ordenada, codigo)

      if encontrado:
          print(f"El producto con código {codigo} fue encontrado.")
      else:
          print(f"El producto con código {codigo} no se encuentra registrado.")


main()
            
      
```
