```python
from collections import deque


# ==========================================
# 1. GRAFO - LISTA DE ADYACENCIA
# ==========================================

grafo = {
    "A": ["B", "C"],
    "B": ["A", "D", "E"],
    "C": ["A", "F"],
    "D": ["B"],
    "E": ["B", "G"],
    "F": ["C", "G"],
    "G": ["E", "F"]
}


# ==========================================
# MOSTRAR EL GRAFO
# ==========================================

def mostrar_grafo(grafo):
    print("\n===== LISTA DE ADYACENCIA =====")

    for centro, conexiones in grafo.items():
        print(f"{centro} -> {', '.join(conexiones)}")


# ==========================================
# 2. BFS
# ==========================================

def bfs(grafo, inicio):
    visitados = set()
    cola = deque()

    # El centro inicial se marca como visitado
    visitados.add(inicio)
    cola.append(inicio)

    recorrido = []
    niveles = {inicio: 0}

    while cola:
        actual = cola.popleft()
        recorrido.append(actual)

        for vecino in grafo[actual]:

            if vecino not in visitados:
                visitados.add(vecino)
                cola.append(vecino)

                # El nivel del vecino es
                # el nivel del actual + 1
                niveles[vecino] = niveles[actual] + 1

    return recorrido, niveles


# ==========================================
# MOSTRAR BFS
# ==========================================

def mostrar_bfs(grafo, inicio):
    recorrido, niveles = bfs(grafo, inicio)

    print("\n===== RECORRIDO BFS =====")
    print(" -> ".join(recorrido))

    print("\nNiveles:")

    niveles_por_numero = {}

    for centro, nivel in niveles.items():
        if nivel not in niveles_por_numero:
            niveles_por_numero[nivel] = []

        niveles_por_numero[nivel].append(centro)

    for nivel in sorted(niveles_por_numero):
        centros = ", ".join(niveles_por_numero[nivel])
        print(f"Nivel {nivel}: {centros}")


# ==========================================
# 3. DFS
# ==========================================

def dfs(grafo, inicio, visitados, recorrido):
    pila = [inicio]

    while pila:
        actual = pila.pop()

        if actual not in visitados:
            visitados.add(actual)
            recorrido.append(actual)

            for vecino in reversed(grafo[actual]):
                if vecino not in visitados:
                    pila.append(vecino)

# ==========================================
# MOSTRAR DFS
# ==========================================

def mostrar_dfs(grafo, inicio):
    visitados = set()
    recorrido = []

    dfs(grafo, inicio, visitados, recorrido)

    print("\n===== RECORRIDO DFS =====")
    print(" -> ".join(recorrido))


# ==========================================
# 5. BUSCAR UN CENTRO UTILIZANDO BFS
# ==========================================

def buscar_bfs(grafo, inicio, objetivo):
    visitados = set()
    cola = deque()

    # Guardaremos también la ruta
    padres = {}

    visitados.add(inicio)
    cola.append(inicio)

    # El inicio no tiene padre
    padres[inicio] = None

    while cola:
        actual = cola.popleft()

        # Si encontramos el centro
        if actual == objetivo:
            break

        for vecino in grafo[actual]:

            if vecino not in visitados:
                visitados.add(vecino)
                cola.append(vecino)

                # Guardamos de dónde venimos
                padres[vecino] = actual

    # Si el objetivo nunca fue visitado
    if objetivo not in visitados:
        return None

    # Construir la ruta desde el objetivo
    # hasta el inicio
    ruta = []
    actual = objetivo

    while actual is not None:
        ruta.append(actual)
        actual = padres[actual]

    # La ruta se construyó al revés,
    # por eso la invertimos
    ruta.reverse()

    return ruta


# ==========================================
# MOSTRAR BÚSQUEDA
# ==========================================

def mostrar_busqueda(grafo):
    print("\n===== BUSCAR CENTRO =====")

    inicio = input("Centro inicial: ").upper()
    objetivo = input("Centro que desea buscar: ").upper()

    if inicio not in grafo:
        print("Error: el centro inicial no existe.")
        return

    if objetivo not in grafo:
        print("Error: el centro que desea buscar no existe.")
        return

    ruta = buscar_bfs(grafo, inicio, objetivo)

    if ruta is None:
        print("No existe una ruta hasta el centro solicitado.")
    else:
        cantidad_conexiones = len(ruta) - 1

        print("Centro encontrado.")
        print("Ruta:", " -> ".join(ruta))
        print("Cantidad de conexiones:", cantidad_conexiones)


# ==========================================
# 6. AGREGAR NUEVO CENTRO
# ==========================================

def agregar_centro(grafo):
    print("\n===== AGREGAR NUEVO CENTRO =====")

    nuevo = input("Ingrese el nuevo centro: ").upper()

    # Verificar si ya existe
    if nuevo in grafo:
        print("Error: ese centro ya existe.")
        return

    # Crear el nuevo centro
    grafo[nuevo] = []

    print("Centro agregado correctamente.")

    print("\nCentros disponibles:")
    print(", ".join(grafo.keys()))

    print("\nAhora puede agregar las conexiones.")
    print("Escriba FIN cuando termine.")

    while True:
        conexion = input(f"Conectar {nuevo} con: ").upper()

        if conexion == "FIN":
            break

        if conexion not in grafo:
            print("Error: ese centro no existe.")
            continue

        if conexion == nuevo:
            print("Un centro no puede conectarse consigo mismo.")
            continue

        # Evitar conexiones repetidas
        if conexion in grafo[nuevo]:
            print("Esa conexión ya existe.")
            continue

        # Como el grafo es NO dirigido,
        # agregamos la conexión en ambos sentidos
        grafo[nuevo].append(conexion)
        grafo[conexion].append(nuevo)

        print(f"Conexión agregada: {nuevo} - {conexion}")

    print("\nCentro y conexiones agregados correctamente.")


# ==========================================
# SELECCIONAR CENTRO INICIAL
# ==========================================

def seleccionar_centro(grafo):
    while True:
        inicio = input(
            "\nIngrese el centro inicial para los recorridos: "
        ).upper()

        if inicio in grafo:
            return inicio

        print("Error: el centro ingresado no existe.")
        print("Centros disponibles:", ", ".join(grafo.keys()))


# ==========================================
# PROGRAMA PRINCIPAL
# ==========================================

def main():

    print("==========================================")
    print("   RED DE DISTRIBUCIÓN DE MENSAJERÍA")
    print("==========================================")

    # Mostrar grafo inicial
    mostrar_grafo(grafo)

    # Recorridos iniciales desde A
    print("\n===== RECORRIDOS INICIALES DESDE A =====")

    mostrar_bfs(grafo, "A")
    mostrar_dfs(grafo, "A")

    # Menú principal
    while True:

        print("\n==========================================")
        print("                 MENÚ")
        print("==========================================")
        print("1. Mostrar lista de adyacencia")
        print("2. Ejecutar BFS")
        print("3. Ejecutar DFS")
        print("4. Buscar un centro utilizando BFS")
        print("5. Agregar nuevo centro y conexiones")
        print("6. Salir")
        print("==========================================")

        opcion = input("Seleccione una opción: ")

        # ----------------------------------
        # Mostrar grafo
        # ----------------------------------

        if opcion == "1":
            mostrar_grafo(grafo)

        # ----------------------------------
        # BFS
        # ----------------------------------

        elif opcion == "2":
            inicio = seleccionar_centro(grafo)
            mostrar_bfs(grafo, inicio)

        # ----------------------------------
        # DFS
        # ----------------------------------

        elif opcion == "3":
            inicio = seleccionar_centro(grafo)
            mostrar_dfs(grafo, inicio)

        # ----------------------------------
        # Buscar centro
        # ----------------------------------

        elif opcion == "4":
            mostrar_busqueda(grafo)

        # ----------------------------------
        # Agregar centro
        # ----------------------------------

        elif opcion == "5":
            agregar_centro(grafo)

            # Mostrar nueva red
            mostrar_grafo(grafo)

            # Seleccionar nuevamente el inicio
            print("\n===== RECORRIDOS DE LA NUEVA RED =====")

            inicio = seleccionar_centro(grafo)

            mostrar_bfs(grafo, inicio)
            mostrar_dfs(grafo, inicio)

        # ----------------------------------
        # Salir
        # ----------------------------------

        elif opcion == "6":
            print("\nPrograma finalizado.")
            break

        else:
            print("Opción inválida. Intente nuevamente.")


# ==========================================
# INICIAR PROGRAMA
# ==========================================

main()
```

```python
# ==========================================
# PRUEBAS PROPIAS
# ==========================================

print("\n" + "=" * 60)
print("     INICIANDO PRUEBAS PROPIAS Y VALIDACIÓN")
print("=" * 60)


# ==========================================
# 1. GRAFO PROPIO DE PRUEBA
# ==========================================

grafo_propio = {
    "A": ["B", "C"],
    "B": ["A", "D", "E"],
    "C": ["A", "F"],
    "D": ["B"],
    "E": ["B", "G"],
    "F": ["C", "G"],
    "G": ["E", "F"]
}

print("\nGrafo utilizado:")
mostrar_grafo(grafo_propio)


# ==========================================
# PRUEBA 1:
# BFS COMENZANDO DESDE B
# ==========================================

recorrido_bfs_b, niveles_b = bfs(grafo_propio, "B")

print("\nPrueba 1: BFS comenzando desde B")
print("Recorrido obtenido:", " -> ".join(recorrido_bfs_b))

bfs_b_esperado = ["B", "A", "D", "E", "C", "G", "F"]

assert recorrido_bfs_b == bfs_b_esperado, (
    f"Fallo en BFS desde B.\n"
    f"Esperado: {bfs_b_esperado}\n"
    f"Obtenido: {recorrido_bfs_b}"
)

print("✔ Prueba 1 Superada: BFS funciona correctamente comenzando desde B.")


# ==========================================
# PRUEBA 2:
# NIVELES DE BFS DESDE B
# ==========================================

niveles_b_esperados = {
    "B": 0,
    "A": 1,
    "D": 1,
    "E": 1,
    "C": 2,
    "G": 2,
    "F": 3
}

assert niveles_b == niveles_b_esperados, (
    f"Fallo en los niveles de BFS desde B.\n"
    f"Esperado: {niveles_b_esperados}\n"
    f"Obtenido: {niveles_b}"
)

print("✔ Prueba 2 Superada: Los niveles de BFS desde B son correctos.")


# ==========================================
# PRUEBA 3:
# DFS COMENZANDO DESDE C
# ==========================================

visitados_c = set()
recorrido_dfs_c = []

dfs(grafo_propio, "C", visitados_c, recorrido_dfs_c)

print("\nPrueba 3: DFS comenzando desde C")
print("Recorrido obtenido:", " -> ".join(recorrido_dfs_c))

# Comprobamos propiedades del recorrido,
# no un orden específico.

assert recorrido_dfs_c[0] == "C", (
    "Fallo en DFS: El recorrido no comienza en C."
)

assert set(recorrido_dfs_c) == set(grafo_propio.keys()), (
    "Fallo en DFS: No se visitaron todos los centros."
)

assert len(recorrido_dfs_c) == len(set(recorrido_dfs_c)), (
    "Fallo en DFS: Se visitó un centro más de una vez."
)

print("✔ Prueba 3 Superada: DFS comienza en C y visita todos los centros una sola vez.")


# ==========================================
# PRUEBA 4:
# BUSCAR UN CENTRO A VARIAS CONEXIONES
# ==========================================

ruta = buscar_bfs(grafo_propio, "B", "F")

print("\nPrueba 4: Buscar F comenzando desde B")
print("Ruta encontrada:", " -> ".join(ruta))

ruta_esperada = ["B", "A", "C", "F"]

assert ruta == ruta_esperada, (
    f"Fallo en búsqueda BFS.\n"
    f"Esperado: {ruta_esperada}\n"
    f"Obtenido: {ruta}"
)

cantidad = len(ruta) - 1

assert cantidad == 3, (
    f"Fallo: La ruta debería tener 3 conexiones, "
    f"pero tiene {cantidad}."
)

print("✔ Prueba 4 Superada: BFS encontró F mediante la ruta más corta.")


# ==========================================
# PRUEBA 5:
# BUSCAR EL MISMO CENTRO DE INICIO
# ==========================================

ruta_mismo_centro = buscar_bfs(grafo_propio, "E", "E")

print("\nPrueba 5: Buscar E comenzando desde E")
print("Ruta encontrada:", " -> ".join(ruta_mismo_centro))

assert ruta_mismo_centro == ["E"], (
    f"Fallo: La ruta debería ser solamente E, "
    f"pero se obtuvo {ruta_mismo_centro}"
)

assert len(ruta_mismo_centro) - 1 == 0, (
    "Fallo: Buscar el centro inicial debería requerir 0 conexiones."
)

print("✔ Prueba 5 Superada: El centro inicial es encontrado correctamente.")


# ==========================================
# PRUEBA 6:
# VERIFICAR QUE BFS NO REPITA CENTROS
# ==========================================

recorrido_bfs_c, niveles_c = bfs(grafo_propio, "C")

print("\nPrueba 6: Verificar centros visitados en BFS")

assert len(recorrido_bfs_c) == len(set(recorrido_bfs_c)), (
    "Fallo en BFS: Se visitó un centro más de una vez."
)

assert len(recorrido_bfs_c) == len(grafo_propio), (
    "Fallo en BFS: No se visitaron todos los centros."
)

print("✔ Prueba 6 Superada: BFS visita cada centro una sola vez.")


# ==========================================
# PRUEBA 7:
# AGREGAR UN CENTRO PROPIO
# ==========================================

grafo_nuevo = {}

# Copiar el grafo para no modificar el original
for centro in grafo_propio:
    grafo_nuevo[centro] = grafo_propio[centro].copy()


# Agregar el nuevo centro H
grafo_nuevo["H"] = []

# Conectar H con D
grafo_nuevo["H"].append("D")
grafo_nuevo["D"].append("H")

# Conectar H con F
grafo_nuevo["H"].append("F")
grafo_nuevo["F"].append("H")

print("\nPrueba 7: Agregar nuevo centro H")

assert "H" in grafo_nuevo, (
    "Fallo: El centro H no fue agregado."
)

assert "H" in grafo_nuevo["D"], (
    "Fallo: La conexión D-H no fue agregada."
)

assert "D" in grafo_nuevo["H"], (
    "Fallo: La conexión H-D no fue agregada."
)

assert "H" in grafo_nuevo["F"], (
    "Fallo: La conexión F-H no fue agregada."
)

assert "F" in grafo_nuevo["H"], (
    "Fallo: La conexión H-F no fue agregada."
)

print("✔ Prueba 7 Superada: El nuevo centro H fue agregado correctamente.")


# ==========================================
# PRUEBA 8:
# RECORRER LA NUEVA RED
# ==========================================

recorrido_nuevo, niveles_nuevo = bfs(grafo_nuevo, "H")

print("\nPrueba 8: BFS desde el nuevo centro H")
print("Recorrido:", " -> ".join(recorrido_nuevo))

assert recorrido_nuevo[0] == "H", (
    "Fallo: BFS no comenzó desde H."
)

assert len(recorrido_nuevo) == len(grafo_nuevo), (
    "Fallo: BFS no recorrió todos los centros de la nueva red."
)

assert len(recorrido_nuevo) == len(set(recorrido_nuevo)), (
    "Fallo: BFS repitió un centro."
)

print("✔ Prueba 8 Superada: BFS funciona correctamente después de agregar H.")


# ==========================================
# PRUEBA 9:
# DFS EN LA NUEVA RED
# ==========================================

visitados_nuevo = set()
recorrido_dfs_nuevo = []

dfs(
    grafo_nuevo,
    "H",
    visitados_nuevo,
    recorrido_dfs_nuevo
)

print("\nPrueba 9: DFS desde el nuevo centro H")
print("Recorrido:", " -> ".join(recorrido_dfs_nuevo))

assert recorrido_dfs_nuevo[0] == "H", (
    "Fallo: DFS no comenzó desde H."
)

assert set(recorrido_dfs_nuevo) == set(grafo_nuevo.keys()), (
    "Fallo: DFS no visitó todos los centros."
)

assert len(recorrido_dfs_nuevo) == len(set(recorrido_dfs_nuevo)), (
    "Fallo: DFS visitó un centro más de una vez."
)

print("✔ Prueba 9 Superada: DFS funciona correctamente después de agregar H.")


# ==========================================
# RESULTADO FINAL
# ==========================================

print("\n" + "=" * 60)
print("   ¡TODAS LAS PRUEBAS PROPIAS PASARON EXITOSAMENTE!")
print("=" * 60)
```

```python
from random import randint
import time
import matplotlib.pyplot as plt


# ==========================================
# GENERAR GRAFO ALEATORIO
# ==========================================

def generar_grafo(n):
    """
    Genera un grafo no dirigido con n centros.

    Cada centro se conecta inicialmente
    con el siguiente para garantizar que
    todos los centros sean alcanzables.
    """

    grafo = {}

    # Crear los centros
    for i in range(n):
        grafo[str(i)] = []

    # Conectar cada centro con el siguiente
    # para garantizar que exista una red conectada
    for i in range(n - 1):
        grafo[str(i)].append(str(i + 1))
        grafo[str(i + 1)].append(str(i))

    # Agregar algunas conexiones adicionales
    # aleatorias
    cantidad_conexiones = n * 2

    for _ in range(cantidad_conexiones):

        a = str(randint(0, n - 1))
        b = str(randint(0, n - 1))

        # Evitar conexiones consigo mismo
        if a == b:
            continue

        # Evitar conexiones repetidas
        if b not in grafo[a]:
            grafo[a].append(b)
            grafo[b].append(a)

    return grafo


# ==========================================
# PRUEBA DE DESEMPEÑO BFS Y DFS
# ==========================================

def bench_recorridos(n, step=100):

    print("Generando datos de rendimiento para BFS y DFS...", end="")

    resultados_bfs = []
    resultados_dfs = []

    for size in range(50, n + 1, step):

        # Generar una red de prueba
        grafo_prueba = generar_grafo(size)

        # ----------------------------------
        # Medir BFS
        # ----------------------------------

        inicio_tiempo = time.perf_counter()

        bfs(grafo_prueba, "0")

        fin_tiempo = time.perf_counter()

        tiempo_bfs = (fin_tiempo - inicio_tiempo) * 1000

        resultados_bfs.append((size, tiempo_bfs))


        # ----------------------------------
        # Medir DFS
        # ----------------------------------

        visitados = set()
        recorrido = []

        inicio_tiempo = time.perf_counter()

        dfs(
            grafo_prueba,
            "0",
            visitados,
            recorrido
        )

        fin_tiempo = time.perf_counter()

        tiempo_dfs = (fin_tiempo - inicio_tiempo) * 1000

        resultados_dfs.append((size, tiempo_dfs))

        print(".", end="")

    print(" ¡Completado!")

    return resultados_bfs, resultados_dfs


# ==========================================
# GRAFICAR COMPARACIÓN
# ==========================================

def dibujar_comparativa(
    resultados_bfs,
    resultados_dfs,
    dimensiones=(10, 6)
):

    x_bfs = [
        size for size, tiempo in resultados_bfs
    ]

    y_bfs = [
        tiempo for size, tiempo in resultados_bfs
    ]

    x_dfs = [
        size for size, tiempo in resultados_dfs
    ]

    y_dfs = [
        tiempo for size, tiempo in resultados_dfs
    ]

    # Verificar que los datos tengan la misma cantidad
    assert len(x_bfs) == len(y_bfs)
    assert len(x_dfs) == len(y_dfs)

    plt.figure(figsize=dimensiones)

    plt.plot(
        x_bfs,
        y_bfs,
        label="BFS",
        color="crimson",
        linewidth=2,
        marker="o"
    )

    plt.plot(
        x_dfs,
        y_dfs,
        label="DFS",
        color="dodgerblue",
        linewidth=2,
        marker="s"
    )

    plt.xlabel(
        "Tamaño de la red (cantidad de centros)",
        fontsize=12
    )

    plt.ylabel(
        "Tiempo de ejecución (milisegundos)",
        fontsize=12
    )

    plt.title(
        "Prueba de desempeño: BFS vs DFS",
        fontsize=14,
        fontweight="bold"
    )

    plt.legend(fontsize=11)

    plt.grid(
        True,
        linestyle="--",
        alpha=0.6
    )

    plt.show()


# ==========================================
# EJECUCIÓN DE LA PRUEBA DE DESEMPEÑO
# ==========================================

tiempos_bfs, tiempos_dfs = bench_recorridos(
    1500,
    step=100
)

dibujar_comparativa(
    tiempos_bfs,
    tiempos_dfs
)
```
